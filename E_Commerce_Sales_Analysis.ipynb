# **E-commerce Sales Analysis** 

## Introduction
This notebook analyzes an e-commerce sales dataset using Python and Pandas.

The analysis focuses on understanding sales performance across different categories, cities, products, and payment methods. Various Pandas operations such as data selection, filtering, sorting, grouping, and aggregation are used to extract meaningful insights from the dataset.

### Objectives

- Load and explore the e-commerce sales dataset
- Select and filter relevant records
- Sort orders based on sales and quantity
- Analyze sales by category, city, product, and payment method
- Calculate important sales statistics
- Identify top-performing products and categories
- Generate meaningful observations from the analysis

## 3. Data Filtering
Filtering allows us to identify records that satisfy specific conditions.

```python
# Orders with sales greater than ₹10,000

high_value_orders = df[df["Total_Sales"] > 10000]

high_value_orders[
    ["Order_ID", "Product", "City", "Total_Sales"]
]
```

```python
# Orders belonging to the Sports category

sports_orders = df[df["Category"] == "Sports"]

sports_orders[
    ["Order_ID", "Product", "Total_Sales"]
]
```

```python
# Orders paid using UPI
upi_orders = df[df["Payment_Method"] == "UPI"]

upi_orders[
    ["Order_ID", "Product", "Total_Sales"]
]
```

## 4. Sorting the Data
The `sort_values()` function is used to arrange records according to one or more columns.

```python
# Top 10 orders by Total Sales
top_orders = df.sort_values(
    by="Total_Sales",
    ascending=False
)

top_orders[
    ["Order_ID", "Product", "Category", "City", "Total_Sales"]
].head(10)
```

## 5. Overall Sales Analysis
The following calculations provide an overview of the overall performance of the e-commerce dataset.

```python
total_sales = df["Total_Sales"].sum()
average_sales = df["Total_Sales"].mean()
highest_sales = df["Total_Sales"].max()
lowest_sales = df["Total_Sales"].min()
total_quantity = df["Quantity"].sum()
number_of_orders = df["Order_ID"].nunique()

print(f"Total Sales       : ₹{total_sales:,.2f}")
print(f"Average Sales     : ₹{average_sales:,.2f}")
print(f"Highest Sale      : ₹{highest_sales:,.2f}")
print(f"Lowest Sale       : ₹{lowest_sales:,.2f}")
print(f"Total Quantity    : {total_quantity}")
print(f"Number of Orders  : {number_of_orders}")
```

## 6. Category-wise Sales Analysis
The `groupby()` function is used to analyze sales performance across different product categories.

```python
category_analysis = df.groupby("Category").agg(
    Orders=("Order_ID", "count"),
    Total_Sales=("Total_Sales", "sum"),
    Average_Sales=("Total_Sales", "mean"),
    Quantity_Sold=("Quantity", "sum")
).sort_values(
    by="Total_Sales",
    ascending=False
)

category_analysis
```

```python
## 7. City-wise Sales Analysis
```

```python
city_analysis = df.groupby("City").agg(
    Orders=("Order_ID", "count"),
    Total_Sales=("Total_Sales", "sum"),
    Average_Sales=("Total_Sales", "mean"),
    Quantity_Sold=("Quantity", "sum")
).sort_values(
    by="Total_Sales",
    ascending=False
)

city_analysis
```

## 7. Product-wise Sales Analysis

```python
product_analysis = df.groupby("Product").agg(
    Orders=("Order_ID", "count"),
    Total_Sales=("Total_Sales", "sum"),
    Average_Sales=("Total_Sales", "mean"),
    Quantity_Sold=("Quantity", "sum")
).sort_values(
    by="Total_Sales",
    ascending=False
)

product_analysis
```

## 8. Payment Analysis

```python
payment_analysis = df.groupby("Payment_Method").agg(
    Orders=("Order_ID", "count"),
    Total_Sales=("Total_Sales", "sum"),
    Average_Sales=("Total_Sales", "mean"),
    Quantity_Sold=("Quantity", "sum")
).sort_values(
    by="Total_Sales",
    ascending=False
)

payment_analysis
```

## 9. Top performing Products

```python
top_products = product_analysis.head(5)

print("Top 5 Products by Total Sales:")
top_products
```

## 10. Top performing Categories

```python
top_categories = category_analysis.sort_values(
    by="Total_Sales",
    ascending=False
)

print("Categories Ranked by Total Sales:")
top_categories
```

Data selection, filtering, sorting, grouping, and aggregation were used to examine sales across categories, cities, products, and payment methods. Overall sales statistics were also calculated to understand the performance of the dataset.
